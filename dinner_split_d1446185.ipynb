測試一

```
# 1. 載入客戶端提供的原始資料
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people_str = "4"

service_rate = 0.10
discount = 100
budget_per_person = 300

# 1. 整理聚餐名稱：去除頭尾空白並統一轉成小寫
clean_event_name = event_name.strip().lower()

# 2. 整理餐費並轉換型別
clean_price = float(meal_amount.replace('NT$', '').replace(',', '').strip())
people = int(people_str)

# 3. 分步計算費用
service_charge = clean_price * service_rate
total_payable = clean_price + service_charge - discount
price_per_person = total_payable / people

# 4. 判斷是否超過預算並存為布林變數
is_over_budget = price_per_person > budget_per_person

# 5. 輸出費用摘要
print(f"聚餐名稱：{clean_event_name}")
print(f"原始餐費：{clean_price:.2f} 元")
print(f"服務費：{service_charge:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_payable:.2f} 元")
print(f"分攤人數：{people} 人")
print(f"每人費用：{price_per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}")
```
